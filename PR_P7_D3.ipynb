```python
import re

def validate_number_plate_regex(plate: str) -> bool:
    """
    Validates a hypothetical number plate format using a regular expression.
    Example formats:
    - 'AB12 CDE' (2 letters, 2 digits, space, 3 letters)
    - 'AB123 CD' (2 letters, 3 digits, space, 2 letters)
    - 'A1 BCD' (1 letter, 1 digit, space, 3 letters)
    """

    pattern = r'^[A-Z]{2,3}[0-9]{2,4}\s?[A-Z]{2,3}$'
    return bool(re.match(pattern, plate))
```

```python
# Test cases
number_plates = [
    "AB12 CDE",   # Valid (2L 2D 1S 3L)
    "ABC123 DEF", # Valid (3L 3D 1S 3L)
    "A12 BCD",    # Invalid (1L at start)
    "AB1234 EF",  # Valid (2L 4D 1S 2L)
    "AB123EF",    # Valid (no space)
    "123 ABC",    # Invalid (starts with digit)
    "ABC 123 DEF",# Invalid (too many letters in second group, or wrong digit count)
    "XYZ9876MN",  # Valid (3L 4D no space 2L)
    "GHIJ123 KL", # Invalid (too many letters in first group)
    "PQ1RST",     # Invalid (missing digit)
    "PQ123 RSTU"  # Invalid (too many letters in second group)
]

print("--- Number Plate Validation Results ---")
for plate in number_plates:
    is_valid = validate_number_plate_regex(plate)
    print(f"'{plate}': {'Valid' if is_valid else 'Invalid'}")


# Example of how to modify the regex for a specific format:
# Let's say a specific country uses format: `LL DD LLL` (e.g., AB 01 CDE)
print("\n--- Specific Format: LL DD LLL ---")
def validate_specific_format(plate: str) -> bool:
    pattern_specific = r'^[A-Z]{2}\s[0-9]{2}\s[A-Z]{3}$'
    return bool(re.match(pattern_specific, plate))

specific_plates = [
    "AB 01 CDE",  # Valid
    "XY 99 ZZZ",  # Valid
    "A B 01 CDE", # Invalid (extra space)
    "AB01 CDE",   # Invalid (missing space)
    "AB 01CD E",  # Invalid (wrong spacing)
    "12 34 FGH"   # Invalid (starts with digits)
]

for plate in specific_plates:
    is_valid = validate_specific_format(plate)
    print(f"'{plate}': {'Valid' if is_valid else 'Invalid'}")
```

```python
import re

def validate_number_plate_grammar(plate: str) -> bool:
    """
    Validates a number plate using a simplified grammar-based approach.
    Hypothetical grammar rules:
    1. Plate must be composed of 3 parts: Region, Year, and Serial.
    2. Parts are separated by spaces.
    3. Region: 2 uppercase letters (e.g., 'MH', 'DL').
    4. Year: 2 digits (e.g., '01', '23').
    5. Serial: 2-3 uppercase letters followed by 3-4 digits (e.g., 'AB1234', 'CDE567').
    Example: 'MH 01 AB1234' or 'DL 23 CDE567'
    """
    parts = plate.split(' ')

    if len(parts) != 3:
        # Rule 1: Must have 3 parts separated by spaces
        return False

    region_part, year_part, serial_part = parts

    # Rule 3: Validate Region (2 uppercase letters)
    if not re.fullmatch(r'^[A-Z]{2}$', region_part):
        return False

    # Rule 4: Validate Year (2 digits)
    if not re.fullmatch(r'^[0-9]{2}$', year_part):
        return False

    # Rule 5: Validate Serial (2-3 letters followed by 3-4 digits)
    if not re.fullmatch(r'^[A-Z]{2,3}[0-9]{3,4}$', serial_part):
        return False

    return True
```

```python
# Test cases for grammar-based validation
grammar_number_plates = [
    "MH 01 AB1234",   # Valid (Region, Year, Serial - 2L 4D)
    "DL 23 CDE567",   # Valid (Region, Year, Serial - 3L 3D)
    "UP 99 FG890",    # Valid (Region, Year, Serial - 2L 3D)
    "KA 12 XYZ1234",  # Valid (Region, Year, Serial - 3L 4D)
    "MH01 AB1234",    # Invalid (missing space)
    "MH 01AB1234",    # Invalid (missing space in serial)
    "MHA 01 AB1234",  # Invalid (Region too long)
    "MH 1 AB1234",    # Invalid (Year too short)
    "MH 01 AB123",    # Invalid (Serial digits too short)
    "MH 01 A12345",   # Invalid (Serial too many digits)
    "MH 01 12345",    # Invalid (Serial starts with digit)
    "MH 01 ABCDE123", # Invalid (Serial letters too many)
    "MH 01 AB 1234",  # Invalid (too many parts)
    "XX 00 ZZ1234"    # Valid
]

print("\n--- Grammar-based Number Plate Validation Results ---")
for plate in grammar_number_plates:
    is_valid = validate_number_plate_grammar(plate)
    print(f"'{plate}': {'Valid' if is_valid else 'Invalid'}")
```

```python
import re

def parse_number_plate_to_tree(plate: str) -> dict or None:
    """
    Attempts to parse a number plate string into a conceptual tree (dictionary) structure.
    Uses the same hypothetical grammar as before: Region, Year, Serial.
    Example: 'MH 01 AB1234' -> {'Region': 'MH', 'Year': '01', 'Serial': 'AB1234'}
    """
    parts = plate.split(' ')
    if len(parts) != 3:
        return None # Malformed structure, cannot parse into tree

    region_part, year_part, serial_part = parts

    # Further break down serial part into letters and digits for deeper tree structure
    serial_match = re.match(r'([A-Z]{2,3})([0-9]{3,4})$', serial_part)
    if serial_match:
        serial_letters = serial_match.group(1)
        serial_digits = serial_match.group(2)
        serial_node = {'letters': serial_letters, 'digits': serial_digits}
    else:
        serial_node = None # Could not parse serial part

    return {
        'plate_string': plate,
        'components': {
            'region': region_part,
            'year': year_part,
            'serial': serial_node
        }
    }

def validate_plate_tree(plate_tree: dict) -> bool:
    """
    Validates the structured number plate tree against defined rules.
    """
    if not plate_tree or 'components' not in plate_tree:
        return False

    components = plate_tree['components']

    # Validate Region: 2 uppercase letters
    if not re.fullmatch(r'^[A-Z]{2}$', components.get('region', '')):
        return False

    # Validate Year: 2 digits
    if not re.fullmatch(r'^[0-9]{2}$', components.get('year', '')):
        return False

    # Validate Serial component (if it was parsed successfully)
    serial_comp = components.get('serial')
    if not serial_comp:
        return False # Serial part was malformed during parsing

    # Validate Serial Letters: 2-3 uppercase letters
    if not re.fullmatch(r'^[A-Z]{2,3}$', serial_comp.get('letters', '')):
        return False

    # Validate Serial Digits: 3-4 digits
    if not re.fullmatch(r'^[0-9]{3,4}$', serial_comp.get('digits', '')):
        return False

    return True
```

```python
# Test cases for tree-based validation (using the same grammar as before)
tree_number_plates = [
    "MH 01 AB1234",   # Valid
    "DL 23 CDE567",   # Valid
    "UP 99 FG890",    # Valid
    "KA 12 XYZ1234",  # Valid
    "MH01 AB1234",    # Invalid (parsing failure - wrong structure)
    "MH 01AB1234",    # Invalid (parsing failure - wrong serial format)
    "MHA 01 AB1234",  # Invalid (validation failure - region too long)
    "MH 1 AB1234",    # Invalid (validation failure - year too short)
    "MH 01 AB123",    # Valid (serial letters 2, digits 3)
    "MH 01 A12345",   # Invalid (validation failure - serial letters too short)
    "MH 01 12345",    # Invalid (parsing failure - serial starts with digit)
    "MH 01 ABCDE123", # Invalid (validation failure - serial letters too many)
    "MH 01 AB 1234",  # Invalid (parsing failure - too many parts)
    "XX 00 ZZ1234"    # Valid
]

print("\n--- Tree-based Number Plate Validation Results ---")
for plate in tree_number_plates:
    parsed_tree = parse_number_plate_to_tree(plate)
    is_valid = validate_plate_tree(parsed_tree)
    print(f"'{plate}': {'Valid' if is_valid else 'Invalid'}")
```
